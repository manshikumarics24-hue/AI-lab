```python

from itertools import product

# Knowledge Base
# P = It is raining
# Q = Roads are wet
# R = Traffic is slow

# Rules:
# 1. P -> Q
# 2. Q -> R
# 3. P

def implies(a, b):
    return not a or b

def knowledge_base(P, Q, R):
    return implies(P, Q) and implies(Q, R) and P

# Query: R (Traffic is slow)
def query(P, Q, R):
    return R

# Check entailment using truth table
def check_entailment():
    symbols = ['P', 'Q', 'R']
    entailed = True

    print("P Q R | KB | Query")
    print("-------------------")

    for values in product([False, True], repeat=3):
        P, Q, R = values

        kb = knowledge_base(P, Q, R)
        q = query(P, Q, R)

        print(P, Q, R, "|", kb, "|", q)

        if kb and not q:
            entailed = False

    if entailed:
        print("\nQuery is ENTAILED by Knowledge Base.")
    else:
        print("\nQuery is NOT ENTAILED by Knowledge Base.")

check_entailment()
```

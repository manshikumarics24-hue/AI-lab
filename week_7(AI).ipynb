```python

def is_variable(x):
    return isinstance(x, str) and x.startswith("?")

def occurs(var, expr, subst):
    expr = apply_substitution(expr, subst)

    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs(var, item, subst) for item in expr)

    return False

def apply_substitution(expr, subst):
    if is_variable(expr) and expr in subst:
        return apply_substitution(subst[expr], subst)

    if isinstance(expr, tuple):
        return tuple(apply_substitution(x, subst) for x in expr)

    return expr

def unify_var(var, expr, subst):
    if var in subst:
        return unify(subst[var], expr, subst)

    if is_variable(expr) and expr in subst:
        return unify(var, subst[expr], subst)

    if occurs(var, expr, subst):
        return None

    new_subst = subst.copy()
    new_subst[var] = expr
    return new_subst

def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    if x == y:
        return subst

    if is_variable(x):
        return unify_var(x, y, subst)

    if is_variable(y):
        return unify_var(y, x, subst)

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y) or x[0] != y[0]:
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None


# Main program
print("UNIFICATION ALGORITHM")

p1 = ("Knows", "?x", "?y")
p2 = ("Knows", "John", "?z")

print("Expression 1:", p1)
print("Expression 2:", p2)

result = unify(p1, p2)

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)
```

```python
import math
import random


def calculate_cost(board):
    """Calculates the total number of conflicting queen pairs (row and diagonal collisions).

    Target cost = 0.
    """
    conflicts = 0
    n = len(board)
    for c1 in range(n):
        for c2 in range(c1 + 1, n):
            r1, r2 = board[c1], board[c2]
            # Row collision
            if r1 == r2:
                conflicts += 1
            # Diagonal collision (|r1 - r2| == |c1 - c2|)
            elif abs(r1 - r2) == abs(c1 - c2):
                conflicts += 1
    return conflicts


def get_random_neighbor(board):
    """Generates a neighbor state by picking a random column and moving its queen

    to a different random row [0, 7].
    """
    neighbor = board.copy()
    col = random.randint(0, 7)
    current_row = neighbor[col]

    # Choose a row different from the current row
    new_row = random.choice([r for r in range(8) if r != current_row])
    neighbor[col] = new_row
    return neighbor


def simulated_annealing(
    initial_board=None,
    initial_temp=100.0,
    cooling_factor=0.99,
    min_temp=0.0001,
    max_iterations=100000,
):
    """Solves the 8-Queens problem using the Simulated Annealing metaheuristic algorithm.

    Returns the solution board, final cost, and total iterations.
    """
    # 1. State Representation: Initial board configuration
    if initial_board is None:
        current_board = [random.randint(0, 7) for _ in range(8)]
    else:
        current_board = initial_board.copy()

    current_cost = calculate_cost(current_board)
    T = initial_temp
    iteration = 0

    while T > min_temp and current_cost > 0 and iteration < max_iterations:
        # 3. Neighbor Generation
        neighbor_board = get_random_neighbor(current_board)
        neighbor_cost = calculate_cost(neighbor_board)

        # 4. Cost Difference & Acceptance Criterion
        # delta_E = neighbor_cost - current_cost
        delta_E = neighbor_cost - current_cost

        if delta_E < 0:
            # Better move: Always accept
            current_board = neighbor_board
            current_cost = neighbor_cost
        else:
            # Worse or equal move: Accept with probability P = e^(-delta_E / T)
            acceptance_prob = math.exp(-delta_E / T)
            if random.random() < acceptance_prob:
                current_board = neighbor_board
                current_cost = neighbor_cost

        # 5. Cooling Schedule
        T *= cooling_factor
        iteration += 1

    return current_board, current_cost, iteration


def print_chessboard(board):
    """Prints a visual 8x8 ASCII representation of the chessboard state."""
    print("  " + " ".join([f"C{c}" for c in range(8)]))
    for row in range(8):
        line = f"R{row} "
        for col in range(8):
            if board[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)


if __name__ == "__main__":
    # Example starting board (e.g., all queens on diagonal as in step-by-step example)
    initial_state = [0, 1, 2, 3, 4, 5, 6, 7]
    print("Initial State:", initial_state)
    print("Initial Conflicts:", calculate_cost(initial_state))

    # Run Simulated Annealing
    solution_board, final_cost, total_iters = simulated_annealing(
        initial_board=initial_state,
        initial_temp=10.0,
        cooling_factor=0.99,
        min_temp=0.0001,
    )

    print("\n--- Final Results ---")
    print(f"Solution Board (1D): {solution_board}")
    print(f"Final Conflicts (Cost): {final_cost}")
    print(f"Total Iterations: {total_iters}")

    print_chessboard(solution_board)
```
